```python
import pandas as pd
import re

file_path = '/content/drive/MyDrive/it_vacancies_2024.csv'

cols_to_read = ["name", "role", "area", "experience", "salary_from", "salary_to", "currency", "gross", "key_skills"]
df = pd.read_csv(file_path, usecols=cols_to_read, dtype={"gross": "string"})
print(f"Исходный размер: {len(df)} строк")

PROGRAMMER = "Программист, разработчик"
PYTHON_PATTERN = r"python|питон|django|fastapi"

def is_my_segment(row):
    role = str(row.get("role", "")).strip()
    title = str(row.get("name", "")).strip().lower()
    if role == PROGRAMMER and re.search(PYTHON_PATTERN, title):
        return True
    return False

mask = df.apply(is_my_segment, axis=1)
df_python = df[mask].copy()
print(f"Размер сегмента 'Python-разработчик': {len(df_python)} строк")

df_python['skills_list'] = df_python['key_skills'].apply(process_skills_string)

for i in range(3):
    print(f"Вакансия: {df_python.iloc[i]['name']}")
    print(f"Исходные навыки: {df_python.iloc[i]['key_skills']}")
    print(f"Очищенные: {df_python.iloc[i]['skills_list']}")
    print("-" * 30)

output_path = '/content/drive/MyDrive/it_vacancies_python_segment.csv'
df_python.to_csv(output_path, index=False)
print(f"Файл сохранен: {output_path}")
```
